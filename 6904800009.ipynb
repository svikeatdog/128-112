```python
def is_valid_score(score):
    """ตรวจสอบว่าคะแนนอยู่ในช่วง 0 ถึง 100 หรือไม่ คืนค่า True หากถูกต้อง และ False หากไม่ถูกต้อง"""
    return 0 <= score <= 100

def to_grade(score):
    """แปลงคะแนนเป็นเกรด (A, B, C, F) หากคะแนนไม่ถูกต้องจะคืนค่าข้อความแจ้งเตือน"""
    if not is_valid_score(score):
        return "Invalid Score"

    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

def print_result(student_id, score):
    """แสดงผลลัพธ์รายงานคะแนนและเกรดของนักศึกษาแต่ละคน"""
    grade = to_grade(score)
    print(f"Student ID: {student_id} | Score: {score} | Grade: {grade}")
```

```python
print_result("6904800009", 85)
print()
print_result("6904800008", 72)
print()
print_result("6904800006", 55)
print()
print_result("6904800005", 105)   # คะแนนไม่ถูกต้อง
```

```python
def safe_average(total, count):
    if count > 0:
        return total / count
    return None

total = 0
count = 0

for i in range(5):
    try:
        x = float(input(f"PM2.5 วัน {i+1}: "))
        total += x
        count += 1
    except:
        pass

avg = safe_average(total, count)

if avg is None:
    print("ไม่มีข้อมูล")
else:
    print("ค่าเฉลี่ย =", avg)
```
